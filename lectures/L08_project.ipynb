## The expansion of the Universe

This project investigates a well-known problem in Cosmology (a branch of astrophysics that studies the origin, structure, evolution, and ultimate fate of the Universe as a whole): **the expansion of the Universe**! 


In 1920, Hubble, from the observation of a special kind of stars known as Cepheid was able to determine their distance and those of the galaxies where they belong. For these galaxies, he was also able to measure the velocity. From these measurments, Hubble found the existence of a linear relationship between the velocity of the galaxies and their distance from us. [[Hubble 1929](https://www.pnas.org/doi/10.1073/pnas.15.3.168)]

![](https://courses.ems.psu.edu/astro801/sites/astro801/files/image/hubbleslaw_plot.jpg)

That is, things at larger distances are moving faster. This means the Universe is expanding! 



Hubble formulated that the velocity of the galaxies, $v$, can be related to the distance, $d$, through the following relationship: $v = H_0 d$, where $H_0$ is known as the Hubble constant.

But, how do we measure distances and velocities in astronomy?

- **Distances**: In real life, they are measured with some physical instrument (e.g., a ruler). However, in astronomy, usually we have to deal with objects that are very far apart (the distance to the Andromeda galaxy, the closest galaxy to the Milky Way, is of the order of $10^{19}$ km), and we have to use a different approach: absolute and apparent brightness. Suppose we have a source of known (i.e., intrinsic) brightness, e.g., a light bulb. If we move the source far away from us and we measure the brightness, since we know the real, intrinsic one, we can estimate the distance at which the source has been moved. We can use this idea also for some astrophysical objects, but we need to know the intrinsic brightness (not an easy task). For some objects, we know it, for example, Cepheid stars and supernovas. We can infer the distance by considering the difference between the intrinsic magnitude (or brightness) and apparent (measured) magnitude (or brightness).


- **Velocities**: When a source emitting light or sound is moving, the wavelength we measure is different compared to the case when the source is still. This is the Doppler effect (e.g., ambulance example in any high school physics book). Specifically, how the wave is modified depends on whether the source is moving away from us (redshifted) or toward us (blue-shifted). By measuring the amount of shift the wave has experienced, we can measure the velocity. This can be exploited with specific electromagnetic emissions coming from the galaxies. From the redshift, we can infer the velocity.

Hubble's linear relationship between the velocity at which galaxies are running away from us and their distance holds just for the local Universe, i.e., for galaxies very near to us. This is due to the fact that variable stars, like Cepheids, are too hard to see when they are very far away.

At high redshifts, cosmology is done with a specific kind of exploding stars called *Type Ia Supernovae*. As for Cepheids, for this kind of source we know the intrinsic brightness, and once we measure their apparent brightness, we can infer the distance.

We know the intrinsic brightness of Type Ia Supernovae since we understand the mechanism by which they are produced.

![](https://upload.wikimedia.org/wikipedia/commons/thumb/c/ce/Progenitor_IA_supernova.svg/1280px-Progenitor_IA_supernova.svg.png)



When they started using Type 1A Supernova for cosmology, they found another mindblowing surprise!

This is a [Nobel prize](https://www.nobelprize.org/prizes/physics/2011/riess/facts/) plot by [Riess et al](https://iopscience.iop.org/article/10.1086/300499):



Here, on the y-axis (top panel), $\mu = m - M$ (the difference between the apparent and intrinsic magnitude) is plotted, which can be directly interpreted as a distance, and on the x-axis, there is $z$, the redshift, which can be interpreted as the velocity. As you can see, at small distances and velocities, the relationship is still linear, but at high redshift, it deviates!
The non-linearity at high redshift implies that the Universe is not only expanding, but it is accelerating! That is, the expansion in the past was slower than the expansion today.

Following the more established and accepted cosmological model of the Universe, [$\Lambda CDM$ ](https://en.wikipedia.org/wiki/Lambda-CDM_model):
How the Universe is expanding (at a constant or accelerated rate) depends on the content of the Universe itself. From the simplest to the more sophisticated observations, we know that the Universe is made up of *ordinary matter* (atoms, particles, etc.), *light* (photons), and *dark matter* (a special kind of matter that does not interact with ordinary matter or light, but interacts through gravity. It is essential to explain the properties of galaxies and is a very active field of research). Each of these components is associated with a density:
- $\Omega_{\rm b}$ the density of ordinary matter in the Universe
- $\Omega_{\gamma}$ the density of photons energy (light) in the Universe
- $\Omega_{\rm m}$ the density of dark matter in the Universe


However, just these three components are not enough to explain the accelerating expansion of the Universe. For this reason, another component, known as *dark energy*, has been introduced. We have no idea how to interpret this quantity, apart from the fact that we need it to explain the accelerated expansion. To this energy, we associate the density $\Omega_{\Lambda}$.

These $\Omega$ parameters also determine the shape of the Universe. We believe that we live in a flat Universe, which implies:

$$\Omega_{\rm b}+\Omega_{\gamma}+\Omega_{\rm m}+\Omega_{\rm \Lambda}=1$$

where $\Omega_{\rm b}$ and $\Omega_{\gamma}$ are very small compared to $\Omega_{\rm m}$ (dark matter) and $\Omega_{\rm \Lambda}$ (dark energy), and thus can be *ignored*!


In $\Lambda CDM$ , the distance modulus is given by (see e.g. [Hogg 2010](https://arxiv.org/abs/astro-ph/9905116)):

$$\mu = 5 \log \left( \frac{c/H_0} {10 {\rm pc}}(1+z)\int_0^z \frac{dz'}{\sqrt{\Omega_{\rm DM} (1+z')^3+\Omega_\Lambda}} \right) $$


*Reminder:* the parameters of the model are
- $H_0$, the Hubble constant
- $\Omega_{\rm m}$ the density of dark matter in the Universe
- $\Omega_\Lambda$ the density of dark energy in the Universe
- (the density of ordinary matter is so small that can be safely neglected). 
- (we are also neglecting the energy of photons, i.e. light)

*Notes*:
- $c$ is the expression above is the speed of light 
- pc in the expression above stands for parsec with 1 pc = 3.086 x 10^16 meters.
- Astronomers like weird units. The Hubble constant has unit of 1/s but it's typically quoted in (km/s)/Mpc. 
- The Omegas are dimensionless.

# Solution

The plan follows the steps in the text and reuses most of what the course has built: likelihood and priors, MCMC with convergence diagnostics, model selection through the evidence, and kernel density estimation for the mock catalogue.

Four models show up, and they differ only in which of $(H_0, \Omega_m, \Omega_\Lambda)$ are free:

| | free | fixed | question it answers |
|---|---|---|---|
| $M_1$ | $H_0$ | $\Omega_m = 1,\ \Omega_\Lambda = 0$ | Einstein-de Sitter: matter only, no dark energy |
| $M_2$ | $H_0, \Omega_m$ | $\Omega_\Lambda = 1 - \Omega_m$ | flat $\Lambda\mathrm{CDM}$ |
| $M_3$ | $H_0, \Omega_m$ | $\Omega_\Lambda = 0$ | no dark energy, curvature allowed |
| $M_4$ | $H_0, \Omega_m, \Omega_\Lambda$ | - | the full model |

So instead of copy-pasting four near-identical likelihoods I write one $\mu(z)$ predictor and build each model's likelihood from a small function saying how its sampling parameters map onto $(H_0, \Omega_m, \Omega_\Lambda)$.

The expression given in the problem statement:

$$\mu = 5\log_{10}\left[\frac{c/H_0}{10\,\mathrm{pc}}(1+z)\int_0^z\frac{dz'}{\sqrt{\Omega_m(1+z')^3+\Omega_\Lambda}}\right],$$



```python
from scipy.integrate import cumulative_trapezoid
from scipy.stats import gaussian_kde
from scipy.optimize import curve_fit
import emcee
import dynesty
from dynesty import utils as dyfunc
import corner

SEED = 42

C_LIGHT = 299792.458  # speed of light [km/s]

z, mu, dmu = np.loadtxt("../datasets/supernovae.txt").T
print(
    f"{len(z)} supernovae, z in [{z.min():.3f}, {z.max():.3f}], "
    f"sigma_mu in [{dmu.min():.3f}, {dmu.max():.3f}]"
)
```

## Step 1: predicting $\mu$, and checking the integrator

The integral $\int_0^z dz'/E(z')$ has no closed form and the likelihood needs it for all 500 supernovae at every step of every chain, so calling `scipy.quad` once per supernova would be hopeless.

The point is that the integral is cumulative: the same integrand at 500 different upper limits. So I evaluate the integrand once on a fine grid in $z$, integrate it cumulatively with the trapezoid rule and interpolate at the data redshifts. One vectorised pass per likelihood call instead of 500 adaptive quadratures.

```python
ZGRID = np.linspace(0.0, z.max() * 1.001, 3000)  # fine grid, fixed once


def mu_model(zv, H0, Om, OL):
    """Distance modulus in the LambdaCDM model of the problem statement.

    H0 in (km/s)/Mpc, so c/H0 comes out in Mpc; since 1 Mpc = 1e6 pc, the
    5*log10(d_L / 10 pc) of the statement becomes 5*log10(d_L/Mpc) + 25.
    """
    E = np.sqrt(Om * (1.0 + ZGRID) ** 3 + OL)
    integral = cumulative_trapezoid(1.0 / E, ZGRID, initial=0.0)
    d_L = (C_LIGHT / H0) * (1.0 + zv) * np.interp(zv, ZGRID, integral)  # Mpc
    return 5.0 * np.log10(d_L) + 25.0
```

```python
# Check against astropy, i.e. somebody else's implementation of the same physics.
from astropy.cosmology import FlatLambdaCDM

err = np.abs(
    mu_model(z, 70.0, 0.3, 0.7) - FlatLambdaCDM(H0=70, Om0=0.3).distmod(z).value
).max()

print(f"max |difference| vs astropy            = {err:.2e} mag")
print(f"smallest measurement error in the data = {dmu.min():.3f} mag")
```

## The four models, written once

`emcee` samples the posterior and needs `log_prior` + `log_likelihood`; `dynesty` samples the prior and needs a prior transform from the unit cube, where a uniform $\mathcal{U}[a,b]$ becomes `a + (b-a)*u`. Each model below is defined by the names of its free parameters, their uniform ranges, and how they expand into $(H_0, \Omega_m, \Omega_\Lambda)$. Everything else is generated from that.

Priors: $H_0 \sim \mathcal{U}[50,100]$ (km/s)/Mpc, which brackets every measured value, and $\Omega_m, \Omega_\Lambda \sim \mathcal{U}[0,1.5]$, positive by physics and generous at the top. All models share them, so the Occam penalties in the Bayes factors later are like for like.

```python
class Model:
    """One cosmological model: its free parameters, their uniform priors, how they
    expand into the (H0, Om, OL) triple mu_model needs, and the data it is fitted to."""

    def __init__(self, name, labels, bounds, expand, data=None):
        self.name = name
        self.labels = labels  # for plots
        self.bounds = np.asarray(bounds)  # (ndim, 2) uniform prior ranges
        self.expand = expand  # theta -> (H0, Om, OL)
        self.ndim = len(labels)
        self.z, self.mu, self.dmu = data if data is not None else (z, mu, dmu)
        # constant term of the Gaussian log-likelihood: it cancels in the Bayes
        # factors, but keeping it makes ln Z the actual evidence
        self.lnorm = -0.5 * np.sum(np.log(2 * np.pi * self.dmu**2))

    def loglike(self, theta):
        H0, Om, OL = self.expand(theta)
        r = (self.mu - mu_model(self.z, H0, Om, OL)) / self.dmu
        return self.lnorm - 0.5 * np.sum(r**2)

    def log_prior(self, theta):
        lo, hi = self.bounds[:, 0], self.bounds[:, 1]
        return 0.0 if np.all((theta > lo) & (theta < hi)) else -np.inf

    def log_posterior(self, theta):
        lp = self.log_prior(theta)
        return -np.inf if not np.isfinite(lp) else lp + self.loglike(theta)

    def ptform(self, u):
        """dynesty samples the unit cube: map it onto the uniform priors."""
        lo, hi = self.bounds[:, 0], self.bounds[:, 1]
        return lo + (hi - lo) * u


H0_RANGE, OM_RANGE, OL_RANGE = (50.0, 100.0), (0.0, 1.5), (0.0, 1.5)

M1 = Model(
    "M1: Einstein-de Sitter", [r"$H_0$"], [H0_RANGE], lambda th: (th[0], 1.0, 0.0)
)
M2 = Model(
    "M2: flat LCDM",
    [r"$H_0$", r"$\Omega_m$"],
    [H0_RANGE, OM_RANGE],
    lambda th: (th[0], th[1], 1.0 - th[1]),
)
M3 = Model(
    "M3: no dark energy",
    [r"$H_0$", r"$\Omega_m$"],
    [H0_RANGE, OM_RANGE],
    lambda th: (th[0], th[1], 0.0),
)
M4 = Model(
    "M4: full 3D",
    [r"$H_0$", r"$\Omega_m$", r"$\Omega_\Lambda$"],
    [H0_RANGE, OM_RANGE, OL_RANGE],
    lambda th: (th[0], th[1], th[2]),
)
```

```python
def run_mcmc(model, start, nwalkers=32, nsteps=8000, plot_trace=True):
    """emcee, with the burn-in / thinning recipe we have used since lecture 5."""
    np.random.seed(SEED)
    p0 = np.asarray(start) + 1e-3 * np.random.randn(nwalkers, model.ndim)
    sampler = emcee.EnsembleSampler(nwalkers, model.ndim, model.log_posterior)
    sampler.run_mcmc(p0, nsteps, progress=True)

    tau = sampler.get_autocorr_time(tol=0)
    burn, thin = int(3 * tau.max()), max(1, int(0.5 * tau.min()))
    flat = sampler.get_chain(discard=burn, thin=thin, flat=True)
    print(
        f"{model.name}: tau = {np.round(tau, 1)}, burn = {burn}, thin = {thin}, "
        f"independent samples = {flat.shape[0]}, "
        f"acceptance = {sampler.acceptance_fraction.mean():.2f}"
    )

    if plot_trace:
        chain = sampler.get_chain()
        fig, axes = plt.subplots(
            model.ndim, figsize=(10, 2.2 * model.ndim), sharex=True, squeeze=False
        )
        for i in range(model.ndim):
            axes[i, 0].plot(chain[:, :, i], "k", alpha=0.25)
            axes[i, 0].set_ylabel(model.labels[i])
            axes[i, 0].axvline(burn, color="r", ls="--")
        axes[-1, 0].set_xlabel("step")
        fig.suptitle(f"Traces, {model.name} (red = burn-in cut)")
        plt.show()
    return flat


def run_nested(model, nlive=500):
    """dynesty: equally-weighted samples, plus ln Z and its uncertainty."""
    sampler = dynesty.NestedSampler(
        model.loglike,
        model.ptform,
        ndim=model.ndim,
        nlive=nlive,
        bound="multi",
        sample="rwalk",
        rstate=np.random.default_rng(SEED),
    )
    sampler.run_nested(dlogz=0.05, print_progress=True)
    res = sampler.results
    eq = dyfunc.resample_equal(res.samples, np.exp(res.logwt - res.logz[-1]))
    print(f"\n{model.name}: ln Z = {res.logz[-1]:.3f} +/- {res.logzerr[-1]:.3f}")
    return eq, res.logz[-1], res.logzerr[-1]


def summarize(samples, labels, title):
    """Median and 68% credible interval."""
    print(title)
    q = np.percentile(samples, [16, 50, 84], axis=0)
    for i, lab in enumerate(labels):
        lo, med, hi = q[0, i], q[1, i], q[2, i]
        print(f"  {lab:<4} = {med:8.3f}  +{hi - med:.3f} / -{med - lo:.3f}")
    return q[1]
```

## Step 2: the one-parameter model ($\Omega_m = 1$, $\Omega_\Lambda = 0$)

A flat, matter-only Universe, the Einstein-de Sitter model that everyone assumed before 1998. Only $H_0$ is free. I fit it with both samplers, mostly as a check that they agree on a problem simple enough to see the answer.

```python
flat_M1 = run_mcmc(M1, [65.0])
med_M1 = summarize(flat_M1, ["H0"], "\nM1, MCMC (median, 68% CI):")
```

```python
eq_M1, logZ_M1, logZ_M1_err = run_nested(M1)
summarize(eq_M1, ["H0"], "M1, nested sampling (median, 68% CI):")

plt.figure(figsize=(7, 4))
plt.hist(flat_M1[:, 0], bins=40, density=True, alpha=0.4, label="MCMC (emcee)")
plt.hist(
    eq_M1[:, 0], bins=40, density=True, histtype="step", lw=2, label="nested (dynesty)"
)
plt.xlabel(r"$H_0$ [(km/s)/Mpc]")
plt.ylabel("posterior density")
plt.title("One-parameter model: the two samplers agree")
plt.legend()
plt.show()
```

## Step 3: the two-parameter flat model ($\Omega_\Lambda = 1 - \Omega_m$)

Still a flat Universe, but now dark energy is allowed: whatever is not matter is $\Lambda$.

```python
flat_M2 = run_mcmc(M2, [65.0, 0.5])
summarize(flat_M2, ["H0", "Om"], "\nM2, MCMC (median, 68% CI):")

eq_M2, logZ_M2, logZ_M2_err = run_nested(M2)
summarize(eq_M2, ["H0", "Om"], "M2, nested sampling (median, 68% CI):")
```

```python
fig = corner.corner(
    flat_M2,
    labels=M2.labels,
    quantiles=[0.16, 0.5, 0.84],
    levels=(0.68, 0.95),
    show_titles=True,
    title_fmt=".3f",
)
fig.suptitle("Flat LCDM: a curved degeneracy, not an ellipse", y=1.02)
plt.show()

s1, s2 = flat_M1[:, 0].std(), flat_M2[:, 0].std()
print(f"H0 uncertainty with Om fixed to 1 (M1): +/- {s1:.2f} (km/s)/Mpc")
print(f"H0 uncertainty with Om free       (M2): +/- {s2:.2f} (km/s)/Mpc")
print(f"the error grows by a factor {s2 / s1:.1f}")
print(f"correlation coefficient between H0 and Om: {np.corrcoef(flat_M2.T)[0, 1]:.3f}")
```

$H_0$ is now measured a few times worse than in step 2. Nothing about the data changed, the question did: in $M_1$ I asserted $\Omega_m = 1$ and that assertion was doing a lot of work, while here I admit I do not know $\Omega_m$ and marginalise over it. A tight error bar that comes from an assumption is not a measurement of anything.

$H_0$ and $\Omega_m$ are strongly anticorrelated, and the corner plot shows the degeneracy is a curved banana rather than a tilted ellipse. The reason is in the formula: $H_0$ sets the overall vertical scale of the Hubble diagram while $\Omega_m$ controls the curvature of $\mu(z)$, and over a limited redshift range the two partly mimic each other. Only the departure from a straight line breaks the degeneracy, which is why one needs supernovae at high $z$ and not just more of them nearby. A covariance matrix cannot express that curvature, which is the first concrete thing the Bayesian treatment buys here, and step 6 comes back to it.

## Step 4: the full three-parameter model, and two Bayes factors

Now all three parameters are free. To answer both questions I also need $M_3$, curvature allowed but no dark energy:

- is the Universe flat? compare $M_2$ (flatness imposed) against $M_4$ (not imposed)
- does dark energy exist? compare $M_4$ ($\Omega_\Lambda$ free) against $M_3$ ($\Omega_\Lambda = 0$)

This is the step MCMC cannot do on its own, since it needs the evidence, so all four runs below are nested sampling.

```python
eq_M3, logZ_M3, logZ_M3_err = run_nested(M3)
summarize(eq_M3, ["H0", "Om"], "M3, nested sampling (median, 68% CI):")

eq_M4, logZ_M4, logZ_M4_err = run_nested(M4)
med_M4 = summarize(eq_M4, ["H0", "Om", "OL"], "M4, nested sampling (median, 68% CI):")
```

```python
fig = corner.corner(
    eq_M4,
    labels=M4.labels,
    quantiles=[0.16, 0.5, 0.84],
    levels=(0.68, 0.95),
    show_titles=True,
    title_fmt=".3f",
)
fig.suptitle("Full 3D model", y=1.02)
plt.show()
```

```python
print("Evidences:")
for nm, lz, le in [
    ("M1 Einstein-de Sitter", logZ_M1, logZ_M1_err),
    ("M2 flat LCDM         ", logZ_M2, logZ_M2_err),
    ("M3 no dark energy    ", logZ_M3, logZ_M3_err),
    ("M4 full 3D           ", logZ_M4, logZ_M4_err),
]:
    print(f"  ln Z ({nm}) = {lz:10.3f} +/- {le:.3f}")

# is the Universe flat?  M2 (flatness imposed) against M4 (not imposed)
lnB_flat = logZ_M2 - logZ_M4
err_flat = np.hypot(logZ_M2_err, logZ_M4_err)
print(
    f"\nIs the Universe flat?   B(M2/M4) = {np.exp(lnB_flat):.2f}"
    f"   [ln B = {lnB_flat:.2f} +/- {err_flat:.2f}]"
)

# does dark energy exist?  M4 (OL free) against M3 (OL = 0)
lnB_de = logZ_M4 - logZ_M3
err_de = np.hypot(logZ_M4_err, logZ_M3_err)
print(
    f"Does dark energy exist? B(M4/M3) = {np.exp(lnB_de):.2f}"
    f"   [ln B = {lnB_de:.2f} +/- {err_de:.2f}]"
)

print(
    f"\n(bonus) flat LCDM vs Einstein-de Sitter: "
    f"B(M2/M1) = {np.exp(logZ_M2 - logZ_M1):.2f}"
)
```

### Reading the two Bayes factors

Is the Universe flat? The Bayes factor is close to one, so the data neither establish nor reject flatness: with 500 supernovae and nothing else, imposing $\Omega_m + \Omega_\Lambda = 1$ is neither rewarded nor punished. A Bayes factor near unity means the data are not informative about this question, which is a different statement from the models being equally good.

Does dark energy exist? Here the Bayes factor is a few, which Jeffreys calls substantial but not more. The data do prefer $\Omega_\Lambda > 0$, and the marginal posterior in $M_4$ keeps $\Omega_\Lambda = 0$ about two standard deviations from the median, but a factor of a few is not much on its own. Real cosmology combines supernovae with the CMB and baryon acoustic oscillations precisely because each probe is degenerate in a different direction.

The ordering of the four evidences is also worth noting. Einstein-de Sitter, the model everyone assumed before 1998, is the worst of the four, and the residual plot shows why: it drifts away from the data at high $z$. Even so the Bayes factor against it is only a few. With 500 supernovae of this quality no single one of these questions is settled on its own, but all four comparisons point the same way: a Universe with some dark energy describes the data better than one without.

## Step 5: ten times more supernovae

The mission question: what would $10\times$ the data buy? I build a mock catalogue with the same character as the real one, in three steps.

1. Redshifts: draw new $z$ distributed like the observed ones, which is the density estimation problem of lecture 3. Fit a KDE to the observed $z$ and resample from it.
2. Uncertainties: $\sigma_\mu$ grows with $z$, since fainter objects are noisier, so I fit a smooth trend $\sigma_\mu(z)$ and add back the observed scatter around it.
3. Measurements: evaluate the fitted model at each mock redshift and add Gaussian noise of that size.

Step 3 is where the caveat lives, and I come back to it after the numbers.

```python
rng = np.random.default_rng(1)
N_CLONE = 10 * len(z)

# 1. density estimate in z, and resampling from it
kde_z = gaussian_kde(z, bw_method="scott")
draw = kde_z.resample(2 * N_CLONE, seed=rng)[0]  # oversample...
draw = draw[(draw > z.min()) & (draw < z.max())]  # ...then drop the KDE tails
z_clone = np.sort(draw[:N_CLONE])
assert z_clone.size == N_CLONE

zg = np.linspace(z.min(), z.max(), 300)
plt.figure(figsize=(8, 4))
plt.hist(z, bins=30, density=True, alpha=0.4, label="observed $z$")
plt.hist(z_clone, bins=60, density=True, histtype="step", lw=2, label="cloned $z$")
plt.plot(zg, kde_z(zg), "C3", lw=2, label="KDE")
plt.xlabel("redshift $z$")
plt.ylabel("density")
plt.title("Step 1: a density estimate of the redshift distribution")
plt.legend()
plt.show()
```

```python
# 2. the uncertainty model: smooth trend in z plus the observed scatter around it
trend = np.poly1d(np.polyfit(z, dmu, 2))
scatter = np.std(dmu - trend(z))
dmu_clone = np.clip(
    trend(z_clone) + rng.normal(0, scatter, N_CLONE), dmu.min(), dmu.max()
)

plt.figure(figsize=(8, 4))
plt.plot(z, dmu, ".k", ms=4, alpha=0.6, label=r"observed $\sigma_\mu$")
plt.plot(
    z_clone, dmu_clone, ".", color="C1", ms=3, alpha=0.3, label=r"cloned $\sigma_\mu$"
)
plt.plot(zg, trend(zg), "C3", lw=2, label="quadratic trend")
plt.xlabel("redshift $z$")
plt.ylabel(r"$\sigma_\mu$")
plt.title("Step 2: how the measurement error grows with redshift")
plt.legend()
plt.show()

# 3. the mock measurements, from the flat model fitted in step 3
H0_fit, Om_fit = np.median(flat_M2, axis=0)
mu_clone = mu_model(z_clone, H0_fit, Om_fit, 1 - Om_fit) + rng.normal(0, dmu_clone)
print(
    f"cloned catalogue: {N_CLONE} supernovae, from H0 = {H0_fit:.2f}, Om = {Om_fit:.3f}"
)

plt.figure(figsize=(9, 5.5))
plt.errorbar(
    z_clone,
    mu_clone,
    dmu_clone,
    fmt=".",
    color="C0",
    ecolor="lightsteelblue",
    ms=3,
    lw=0.4,
    alpha=0.5,
    label=f"cloned ({N_CLONE} SNe)",
)
plt.errorbar(z, mu, dmu, fmt=".k", ecolor="gray", lw=1, label=f"real ({len(z)} SNe)")
plt.plot(
    z_clone,
    mu_model(z_clone, H0_fit, Om_fit, 1 - Om_fit),
    "C3",
    lw=2,
    label="model used for cloning",
)
plt.xlabel("redshift $z$")
plt.ylabel(r"$\mu$")
plt.title("Step 3: the mock catalogue")
plt.legend(loc="lower right")
plt.show()
```

```python
# Refit the flat model on the cloned catalogue: same Model, different data.
M2_clone = Model(
    "M2 on the cloned catalogue",
    M2.labels,
    M2.bounds,
    M2.expand,
    data=(z_clone, mu_clone, dmu_clone),
)
flat_clone = run_mcmc(M2_clone, [H0_fit, Om_fit], plot_trace=False)
summarize(flat_clone, ["H0", "Om"], "\nFlat model on 10x the data (median, 68% CI):")

print("\nHow much did the uncertainties shrink?")
for i, lab in enumerate(["H0", "Om"]):
    a, b = flat_M2[:, i].std(), flat_clone[:, i].std()
    print(f"  {lab:<4}: {a:.3f}  ->  {b:.3f}   (factor {a / b:.2f})")
print(f"  expected for 10x the data: sqrt(10) = {np.sqrt(10):.2f}")

# same axis ranges for both, otherwise the two corner plots cannot be overlaid
both = np.vstack([flat_M2, flat_clone])
ranges = list(zip(both.min(axis=0), both.max(axis=0)))
fig = corner.corner(
    flat_M2,
    labels=M2.labels,
    levels=(0.68, 0.95),
    range=ranges,
    color="k",
    plot_datapoints=False,
    hist_kwargs={"density": True},
)
corner.corner(
    flat_clone,
    fig=fig,
    levels=(0.68, 0.95),
    range=ranges,
    color="C3",
    plot_datapoints=False,
    hist_kwargs={"density": True},
)
fig.suptitle("500 supernovae (black) vs 5000 (red)", y=1.02)
plt.show()
```

The uncertainties shrink by roughly $\sqrt{10} \approx 3.2$, which is what independent measurements should do: the Fisher information adds up linearly with the number of data points, so errors fall as $1/\sqrt{N}$. So the answer to ESA is that ten times the supernovae, with the same instrument and the same redshift coverage, would measure $H_0$ and $\Omega_m$ about three times better.

Two caveats, because this kind of forecast is easy to oversell. The clone can forecast but it cannot test: I generated the mock $\mu$ from the flat model I had just fitted, so of course that model fits the clone beautifully. The exercise is about precision, not accuracy. And $1/\sqrt{N}$ eventually stops, because it assumes the errors stay purely statistical and independent, while in a real survey calibration, dust and the standardisation of the supernovae are systematic and do not average down. Past some $N$ the error bar stops improving, which is why real missions spend their effort on controlling systematics and on extending the redshift range, which breaks degeneracies, rather than on raw counts.

## Step 6: forget Bayes for a moment

A non-Bayesian fit for comparison: Levenberg-Marquardt least squares (`scipy.optimize.curve_fit`) on the same two-parameter flat model. It finds the minimum of $\chi^2$ and approximates the uncertainty from the curvature of $\chi^2$ at that minimum, so it returns a point estimate and a covariance matrix rather than a distribution.


```python
def model_flat(zv, H0, Om):
    """The flat two-parameter model, in the signature curve_fit expects."""
    return mu_model(zv, H0, Om, 1.0 - Om)


popt, pcov = curve_fit(
    model_flat, z, mu, p0=[65.0, 0.5], sigma=dmu, absolute_sigma=True
)
perr = np.sqrt(np.diag(pcov))
chi2 = np.sum(((mu - model_flat(z, *popt)) / dmu) ** 2)

print("Levenberg-Marquardt (curve_fit):")
print(f"  H0 = {popt[0]:.2f} +/- {perr[0]:.2f}")
print(f"  Om = {popt[1]:.3f} +/- {perr[1]:.3f}")
print(f"  chi2/dof = {chi2 / (len(z) - 2):.3f}")

print("MCMC on the same model:")
print(f"  H0 = {flat_M2[:, 0].mean():.2f} +/- {flat_M2[:, 0].std():.2f}")
print(f"  Om = {flat_M2[:, 1].mean():.3f} +/- {flat_M2[:, 1].std():.3f}")

fig = corner.corner(
    flat_M2,
    labels=M2.labels,
    quantiles=[0.16, 0.5, 0.84],
    levels=(0.68, 0.95),
    show_titles=True,
    title_fmt=".3f",
)
corner.overplot_lines(fig, popt, color="C3")
fig.suptitle("MCMC posterior (black) vs Levenberg-Marquardt (red)", y=1.02)
plt.show()
```

### Comment: what Bayesian statistics bought, and what it cost

The first thing to say is that the methods agree. The Levenberg-Marquardt fit lands essentially on the posterior median, with comparable error bars, in a fraction of a second instead of minutes. For a well-behaved problem with plenty of data and a single likelihood peak that is what should happen: least squares is the maximum-likelihood estimate, and the Gaussian it assumes around the optimum describes a nearly Gaussian posterior well. It would be dishonest to pretend the Bayesian machinery was necessary just to measure $H_0$ here.

What the Bayesian treatment added:

- the shape of the uncertainty and not only its size. The $H_0$-$\Omega_m$ degeneracy is a curved banana and a covariance matrix can only describe an ellipse, which matters as soon as this measurement is combined with another one
- marginalisation. The error on $H_0$ in step 3 already includes the ignorance of $\Omega_m$, by construction
- model comparison. Step 4 does not exist in the least-squares world, since $\chi^2$ always improves when a parameter is added, while the evidence weighs goodness of fit against prior volume and penalises flexibility on its own. This is the biggest gain and the reason the project needed nested sampling
- explicit priors. $\Omega_m > 0$ is a physical statement, written where a reader can disagree with it instead of hidden in an optimiser's bounds
- no asymptotics. A credible interval means what it says at any sample size, while least-squares error bars degrade when the data are few or the model very non-linear

What it cost: minutes instead of milliseconds, multiplied by the number of models, which is irrelevant for a one-off fit and decisive inside a pipeline; prior ranges that are a real choice, since the Bayes factors of step 4 depend on the assumed prior volume of $\Omega_\Lambda$ and a referee is entitled to argue about it; and more ways to be wrong quietly, since an unconverged chain or too few live points still produce a plausible-looking posterior, which is why every fit above came with trace plots, autocorrelation times and a second sampler.

Choosing a model is what makes an inference mean something, and the Bayesian framework is the one that also lets you ask whether that choice was a good one.
